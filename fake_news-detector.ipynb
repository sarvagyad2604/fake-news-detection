```python
import pandas as pd
import numpy as np
```

```python
true = pd.read_csv('True.csv')
fake = pd.read_csv('Fake.csv')
```

```python
true['label']=1
```

```python
fake['label']=0
```

```python
news = pd.concat([fake,true],axis=0)
```

```python
news=news.drop(['title','subject','date'],axis=1)
```

```python
news=news.sample(frac=1)
```

```python
news.reset_index(inplace=True)
```

```python
news=news.drop(['index'],axis=1)
```

```python
news.head()
```

```python
import re## regular expression
```

```python
def wordopt(text):
  #convert to lowercase
  text=text.lower()

  #remove urls
  text=re.sub(r'https?://\S+|www\.\S+','',text)

  #remove html tags
  text=re.sub('<.*?>+','',text)
  #remove punctuations
  text=re.sub(r'[^\w\s]','',text)
  #remove digits
  text=re.sub(r'\d+','',text)
  #remove extra spaces
  text=re.sub(r'\s+',' ',text)
  #remove newline characters
  text=re.sub(r'\n','',text)
  return text
```

```python
news['text']=news['text'].apply(wordopt)
```

```python
news['text']
```

```python
x = news['text']
y=news['label']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vector=TfidfVectorizer()
```

```python
xv_train=vector.fit_transform(x_train)
xv_test=vector.transform(x_test)
```

```python
xv_train
```

```python
xv_test
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
LR=LogisticRegression()
LR.fit(xv_train,y_train)
```

```python
pred_lr=LR.predict(xv_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, pred_lr)
print(f"Accuracy: {accuracy}")

report = classification_report(y_test, pred_lr)
print("Classification Report:")
print(report)
```

```python
LR.score(xv_test,y_test)
```

```python
print(classification_report(y_test,pred_lr))
```

```python
from sklearn.tree import DecisionTreeClassifier
DT=DecisionTreeClassifier()
DT.fit(xv_train,y_train)
```

```python
pred_dt=DT.predict(xv_test)
```

```python
DT.score(xv_test,y_test)
```

```python
print(classification_report(y_test,pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier
RF=RandomForestClassifier()
RF.fit(xv_train,y_train)
```

```python
pred_rfc=RF.predict(xv_test)
```

```python
RF.score(xv_test,y_test)
```

```python
print(classification_report(y_test,pred_rfc))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
GB=GradientBoostingClassifier()
GB.fit(xv_train,y_train)
```

```python
pred_gb=GB.predict(xv_test)
```

```python
GB.score(xv_test,y_test)
```

```python
def output_label(n):
  if n ==0:
    return "it is a fake news"
  elif n==1:
    return "it's a true news"
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Logistic Regression Performance:")
print(f"Accuracy: {accuracy_score(y_test, pred_lr)}")
print("Classification Report:")
print(classification_report(y_test, pred_lr))

print("\nDecision Tree Classifier Performance:")
print(f"Accuracy: {accuracy_score(y_test, pred_dt)}")
print("Classification Report:")
print(classification_report(y_test, pred_dt))

print("\nRandom Forest Classifier Performance:")
print(f"Accuracy: {accuracy_score(y_test, pred_rfc)}")
print("Classification Report:")
print(classification_report(y_test, pred_rfc))

print("\nGradient Boosting Classifier Performance:")
print(f"Accuracy: {accuracy_score(y_test, pred_gb)}")
print("Classification Report:")
print(classification_report(y_test, pred_gb))
```

```python
def manual_testing(news):
  testing_news={'text':[news]}
  new_def_test=pd.DataFrame(testing_news)
  new_def_test["text"]=new_def_test["text"].apply(wordopt)
  new_x_test=new_def_test["text"]
  new_xv_test=vector.transform(new_x_test)
  pred_LR=LR.predict(new_xv_test)
  pred_DT=DT.predict(new_xv_test)
  pred_RFC=RF.predict(new_xv_test)
  pred_GB=GB.predict(new_xv_test)
  return print("\n\nLR Prediction: {} \nDT Prediction: {} \nRFC Prediction: {} \nGB Prediction: {}".format(output_label(pred_LR[0]),
                                                                                                       output_label(pred_DT[0]),
                                                                                                       output_label(pred_RFC[0]),
                                                                                                       output_label(pred_GB[0])))
```

```python
news_article = str(input())

manual_testing(news_article)
```
